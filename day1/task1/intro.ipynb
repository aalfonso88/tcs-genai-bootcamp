### Test chat completion model

```python
import ollama
```

```python
mensaje = """Que me sugieres hacer en un dia frio?"""
llm = "llama3.2"
```

```python
response = ollama.chat(model=llm, messages=[
    {
        'role': 'user',
        'content': mensaje
    }
])
```

```python
print(response['message']['content'])
```

### Test embedding model

```python
import ollama
```

```python
long = ollama.embeddings(model="nomic-embed-text", prompt="el perro duerme en el parque")
short = ollama.embeddings(model="nomic-embed-text", prompt="perro")
```

```python
print(f"len(long) = {len(long.embedding)}", f"len(short) = {len(short.embedding)}")
```
