```python
import ollama
```

#### ZERO SHOT

```python
mensaje = """¿Esta frase expresa una opinión positiva sobre el producto? 'El producto funciona perfectamente.'"""
llm = 'llama3.2'
```

```python
response = ollama.chat(
    model=llm,
    messages=[
        {
            'role': 'user',
            'content': mensaje
        }
    ]
)
```

```python
print(response['message']['content'])
```

```python
response = ollama.chat(
    model=llm,
    messages=[
        {
            'role': 'user',
            'content': mensaje
        }
    ]
)
```
