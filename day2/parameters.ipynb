#### Conclusión respecto de TOP-P


> top_p conserva los candidatos necesarios para acumular una probabilidad de aproximadamente p.\
> Con un p bajo (0.1 = 10%) el candidato tiende a ser uno solo (el primero) y por lo tanto coincide con k=1\
> Con p más grande (0.9) los candidatos son más variados y las respuestas más variadas.

#### Conclusiones

| Parámetro       | Bajo                 | Alto                  | ¿Qué controla?                         |
| --------------- | -------------------- | --------------------- | -------------------------------------- |
| **Temperatura** | Más determinista     | Más variado/creativo  | Cómo se distribuyen las probabilidades |
| **Top-k**       | Pocos candidatos     | Más candidatos        | **Cantidad** de tokens considerados    |
| **Top-p**       | Poca prob. acumulada | Mayor prob. acumulada | **Masa de probabilidad** considerada   |


#### MAX_TOKENS / NUM_PREDICT

Límite de tokens de salida: determina la cantidad máxima de tokens que el modelo puede generar en su respuesta; \
un límite demasiado bajo puede hacer que la respuesta quede truncada.

```python
mensaje = """Explica en un párrafo que es un embedding en IA"""
def get_response(max_t):
    response = ollama.chat(
        model=llm,
        messages=[
            {
                'role': 'user',
                'content': mensaje
            }
        ],
        options={
            'temperature': 1.0,
            'num_predict': max_t
        }
    )
    return response['message']['content']
```

```python
print(get_response(max_t=20))
```

```python
print(get_response(max_t=200))
```
